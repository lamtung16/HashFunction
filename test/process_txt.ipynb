```python
import csv
import re

input_file = "input.txt"
output_file = "output.csv"

seen = set()  # Track unique rows to prevent duplicates

with open(input_file, "r", encoding="utf-8") as f, open(
    output_file, "w", newline="", encoding="utf-8"
) as out:
  writer = csv.writer(out)

  # CSV header for the 3 requested columns
  writer.writerow(["host", "username", "password"])

  for line in f:
    line = line.strip()
    if not line:
      continue

    # Split by 2+ spaces
    parts = re.split(r"\s{2,}", line)

    try:
      # Remove all spaces from each field
      host = parts[0].replace(" ", "")
      username, password = parts[1].split(":", 1)
      username = username.replace(" ", "")
      password = password.replace(" ", "")

      row = (host, username, password)  # Tuple used so it can be added to the set

      if row not in seen:
        seen.add(row)
        writer.writerow(row)

    except Exception as e:
      print("Skipped line:", line)
```

```python
import re
import csv

input_file = "input.txt"
output_file = "output.csv"

pattern = re.compile(
    r"├─𝙷𝚘𝚜𝚝 = (?P<host>.+?)\n"
    r"├─𝚁𝚎𝚊𝚕 = .+\n"
    r"├─𝙿𝚘𝚛𝚝 = .+\n"
    r"├─𝚄𝚜𝚎𝚛 = (?P<username>.+?)\n"
    r"├─𝙿𝚊𝚜𝚜 = (?P<password>.+?)\n"
    r"├─𝙴𝚡𝚙\.?\s*= .+\n"
    r"├─𝙰𝚌𝚝\s*𝙲𝚘𝚗 = (?P<active_cons>.+?)\n"
    r"├─𝙼𝚊𝚡\s*𝙲𝚘𝚗 = (?P<max_connections>.+?)",
    re.MULTILINE
)

with open(input_file, "r", encoding="utf-8") as f:
    data = f.read()

matches = pattern.finditer(data)

rows = []
for m in matches:
    rows.append([
        m.group("host"),
        m.group("username"),
        m.group("password"),
        m.group("active_cons"),
        m.group("max_connections")
    ])

with open(output_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["host", "username", "password", "active_cons", "max_connections"])
    writer.writerows(rows)

print(f"Exported {len(rows)} rows to {output_file}")
```

```python
import csv
import re

input_file = "input.txt"
output_file = "output.csv"

# Regex to capture host, username, password
pattern = re.compile(r"(https?://[^/]+)/get\.php\?username=([^&]+)&password=([^&]+)")

rows = []

with open(input_file, "r", encoding="utf-8", errors="ignore") as f:
    for line in f:
        line = line.strip()
        match = pattern.search(line)
        if match:
            host = match.group(1)
            username = match.group(2)
            password = match.group(3)
            rows.append([host, username, password])

# Write to CSV
with open(output_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["host", "username", "password"])
    writer.writerows(rows)

print(f"Saved {len(rows)} entries to {output_file}")
```

```python
import csv
import re

input_file = "input.txt"
output_file = "output.csv"

data = []
current_host = None

with open(input_file, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()

        # Match Portal line
        portal_match = re.match(r"Portal:\s*(.+)", line)
        if portal_match:
            current_host = portal_match.group(1)
            continue

        # Match Username/Password line
        cred_match = re.match(r"Username:\s*(.+?)\s*\|\s*Password:\s*(.+)", line)
        if cred_match and current_host:
            username = cred_match.group(1)
            password = cred_match.group(2)
            data.append([current_host, username, password])

# Write CSV
with open(output_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["host", "username", "password"])
    writer.writerows(data)

print(f"Saved {len(data)} records to {output_file}")
```
